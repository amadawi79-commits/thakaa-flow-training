## Task 3
تنظيف البيانات وتوثيق أسباب القرارات

```python
# نسخة مستقلة للتنظيف مع الاحتفاظ بالبيانات الأصلية
df_clean = df.copy()

for column in ["Gender", "Married", "Self_Employed"]:
    missing_count = df_clean[column].isna().sum()
    fill_value = df_clean[column].mode()[0]

    print(column)
    print("عدد القيم الناقصة:", missing_count)
    print("الفئة الأكثر تكرارًا:", fill_value)

    df_clean[column] = df_clean[column].fillna(fill_value)
    print("النقص بعد التعويض:", df_clean[column].isna().sum())
    print()
```

```python
fill_value = df_clean["Dependents"].mode()[0]

print("عدد القيم الناقصة:", df_clean["Dependents"].isna().sum())
print("الفئة الأكثر تكرارًا:", fill_value)
print("عدد قيم 3+:", (df_clean["Dependents"] == "3+").sum())

df_clean["Dependents"] = (
    df_clean["Dependents"]
    .fillna(fill_value)
    .replace("3+", "3")
    .astype(int)
)

print("النقص بعد التعويض:", df_clean["Dependents"].isna().sum())
print("نوع العمود:", df_clean["Dependents"].dtype)
display(df_clean["Dependents"].value_counts())
```

```python
print("عدد القيم الناقصة:", df_clean["LoanAmount"].isna().sum())
print("المتوسط:", round(df_clean["LoanAmount"].mean(), 2))

loan_median = df_clean["LoanAmount"].median()
print("الوسيط المستخدم للتعويض:", loan_median)

df_clean["LoanAmount"] = df_clean["LoanAmount"].fillna(loan_median)

print("النقص بعد التعويض:", df_clean["LoanAmount"].isna().sum())
```

```python
term_mode = df_clean["Loan_Amount_Term"].mode()[0]

print("عدد القيم الناقصة:", df_clean["Loan_Amount_Term"].isna().sum())
print("المدة الأكثر تكرارًا بالأشهر:", term_mode)
print("المدة بالسنوات:", term_mode / 12)

df_clean["Loan_Amount_Term"] = (
    df_clean["Loan_Amount_Term"].fillna(term_mode)
)

print("النقص بعد التعويض:", df_clean["Loan_Amount_Term"].isna().sum())
```

```python
credit_check = pd.DataFrame({
    "credit_group": df["Credit_History"].fillna(-1),
    "approved": df["Loan_Status"].map({"Y": 1, "N": 0})
})

credit_summary = credit_check.groupby("credit_group").agg(
    requests=("approved", "size"),
    approval_percent=("approved", "mean")
)

credit_summary["approval_percent"] = (
    credit_summary["approval_percent"] * 100
).round(2)

display(credit_summary)
```

### قرار معالجة نقص السجل الائتماني

نسبة القبول للطلبات ذات السجل الناقص 74%، وهي أقرب للفئة 1
التي بلغت 79.58%، مقارنة بـ7.87% للفئة 0.
لكن تشابه نسب القبول لا يثبت قيمة السجل المفقودة.
لذلك سأعوّض القيم الخمسين الناقصة بالرمز -1 بمعنى «غير معروف»،
للاحتفاظ بالطلبات دون افتراض سجل ائتماني غير مؤكد.

```python
print("عدد القيم الناقصة:", df_clean["Credit_History"].isna().sum())

df_clean["Credit_History"] = (
    df_clean["Credit_History"].fillna(-1).astype(int)
)

print("النقص بعد المعالجة:", df_clean["Credit_History"].isna().sum())
display(df_clean["Credit_History"].value_counts())
```

```python
outlier_results = []

for column in ["ApplicantIncome", "CoapplicantIncome", "LoanAmount"]:
    q1 = df_clean[column].quantile(0.25)
    q3 = df_clean[column].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    outliers = (
        (df_clean[column] < lower) |
        (df_clean[column] > upper)
    )

    outlier_results.append({
        "column": column,
        "lower_limit": lower,
        "upper_limit": upper,
        "outlier_count": int(outliers.sum())
    })

outlier_summary = pd.DataFrame(outlier_results)
display(outlier_summary.round(2))
```

### قرار القيم المتطرفة

بعد تعويض النقص، كشف فحص IQR عن 50 قيمة متطرفة في دخل المتقدم،
و18 في دخل الشريك، و41 في مبلغ القرض.
قررت الاحتفاظ بها لعدم وجود دليل يؤكد أنها أخطاء؛
فقد تمثل دخولًا أو قروضًا مرتفعة فعلًا، وحذفها قد يفقدنا معلومات مهمة.
هذه أعداد لكل عمود، وقد يتكرر الطلب نفسه في أكثر من عمود.

```python
print("قبل التحويل:")
display(df_clean["Loan_Status"].value_counts(dropna=False))

df_clean["Loan_Status"] = df_clean["Loan_Status"].replace({
    "Y": 1,
    "N": 0
}).astype(int)

print("بعد التحويل:")
display(df_clean["Loan_Status"].value_counts(dropna=False))
```

## Task 5
تنظيف سجلات متحف الديناصورات

```python
fossils = {
    "fossil_id": [
        "F01", "F02", "F03", "F04", "F05",
        "F02", "F06", "F07", "F08", "F09"
    ],
    "species": [
        "T-Rex", "Triceratops", "t-rex", "Velociraptor",
        "Stegosaurus", "Triceratops", "Brachiosaurus",
        "T-Rex", "Velociraptor", "Stegosaurus"
    ],
    "age_million_years": [
        68, 67, 66, 75, -150, 67, 150, 68, None, 155
    ],
    "length": [
        "12 m", "9 m", "40 ft", "2 m", "9 m",
        "9 m", "26 m", "12.3 m", "6.6 ft", "9 m"
    ],
    "found_in": [
        "USA", "USA", "Canada", "Mongolia", "USA",
        "USA", "Tanzania", "USA", "Mongolia", "Portugal"
    ]
}

fossils_df = pd.DataFrame(fossils)
display(fossils_df)
```

```python
fossils_clean = fossils_df.copy()

print("عدد الصفوف المكررة:", fossils_clean.duplicated().sum())

fossils_clean = fossils_clean.drop_duplicates().copy()

fossils_clean["species"] = (
    fossils_clean["species"]
    .str.strip()
    .str.lower()
    .replace({"t-rex": "T-Rex"})
)

print("عدد الصفوف بعد حذف التكرار:", len(fossils_clean))
print("أسماء الأنواع بعد التوحيد:")
display(fossils_clean["species"].value_counts())
```

```python
print("عدد الأعمار السالبة:",
      (fossils_clean["age_million_years"] < 0).sum())

# نفترض في هذا التمرين أن السالب خطأ في إدخال الإشارة
fossils_clean["age_million_years"] = (
    fossils_clean["age_million_years"].abs()
)

print("عدد الأعمار الناقصة:",
      fossils_clean["age_million_years"].isna().sum())

species_median = fossils_clean.groupby("species")[
    "age_million_years"
].transform("median")

fossils_clean["age_million_years"] = (
    fossils_clean["age_million_years"].fillna(species_median)
)

display(
    fossils_clean[["fossil_id", "species", "age_million_years"]]
)
```

```python
length_parts = fossils_clean["length"].str.split(expand=True)

length_value = length_parts[0].astype(float)
length_unit = length_parts[1]

conversion = length_unit.map({
    "m": 1,
    "ft": 0.3048
})

fossils_clean["length_m"] = length_value * conversion

display(
    fossils_clean[["fossil_id", "length", "length_m"]]
)
```

```python
print("متوسط طول كل نوع بالمتر:")
display(
    fossils_clean.groupby("species")["length_m"]
    .mean()
    .round(2)
    .to_frame("average_length_m")
)

print("أقدم نوع حسب أعلى عمر مسجل:")
oldest_age = fossils_clean["age_million_years"].max()
display(
    fossils_clean.loc[
        fossils_clean["age_million_years"] == oldest_age,
        ["species", "age_million_years"]
    ]
)

print("عدد الحفريات في كل دولة:")
display(fossils_clean["found_in"].value_counts())
```
