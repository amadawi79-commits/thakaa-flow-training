## Task 1
قراءة الكود وتوقّع النتائج قبل تشغيله

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Sara", "Noura", "Reem", "Hind"],
    "city": ["Riyadh", "Jeddah", "Riyadh", "Dammam"],
    "score": [88, 92, 75, 60]
}, index=["a", "b", "c", "d"])

print(type(df["score"]))
print(type(df[["score"]]))
print(df.loc["b", "city"])
print(df.iloc[1, 1])
print(df.loc["a":"c", "score"].tolist())
print(df.iloc[0:2, 2].tolist())
print(df[df["score"] > 80]["name"].tolist())
print(df["city"].value_counts()["Riyadh"])
```

### ملاحظاتي بعد التشغيل
- df["score"] يعيد Series، أي عمودًا واحدًا.
- df[["score"]] يعيد DataFrame، أي جدولًا بعمود واحد.
- loc يختار بأسماء الصفوف والأعمدة، وiloc يختار بمواقعها العددية التي تبدأ من صفر.
- loc["a":"c"] يشمل الصف c، لذلك أعاد 3 صفوف.
- iloc[0:2] لا يشمل الموقع 2، لذلك أعاد صفين.
- صاحبتا الدرجات الأعلى من 80 هما Sara وNoura.
- مدينة Riyadh ظهرت مرتين.
### الفرق بين NumPy وPandas
- NumPy array مصفوفة مناسبة للحسابات العددية، وعناصرها تكون من نوع بيانات واحد.
- DataFrame جدول له أسماء أعمدة وفهرس للصفوف، ويمكن أن تختلف أنواع البيانات بين أعمدته.
- نحتاج Pandas لأنها تسهّل قراءة الجداول وتنظيفها، ومعالجة القيم الناقصة، والفلترة والتجميع.

### مثال من حياتي: مصروف البيت
- كل صف يمثل عملية شراء.
- الأعمدة: التاريخ، اسم الصنف، التصنيف، الكمية، والسعر.
- قد يكون السعر ناقصًا إذا لم أحتفظ بالفاتورة أو نسيت تسجيله.
### جدول النتائج بعد التشغيل
لم أسجل توقعاتي قبل التشغيل، لذلك أوثّق النتائج الفعلية وأسبابها هنا.

| السطر | الناتج الفعلي | السبب |
|---|---|---|
| type(df["score"]) | Series | اختيار عمود بأقواس مفردة |
| type(df[["score"]]) | DataFrame | اختيار عمود بأقواس مزدوجة يحافظ على شكل الجدول |
| df.loc["b", "city"] | Jeddah | مدينة الصف الذي اسمه b |
| df.iloc[1, 1] | Jeddah | الصف الثاني والعمود الثاني |
| df.loc["a":"c", "score"].tolist() | [88, 92, 75] | loc يشمل نهاية النطاق c |
| df.iloc[0:2, 2].tolist() | [88, 92] | iloc يستبعد نهاية النطاق 2 |
| df[df["score"] > 80]["name"].tolist() | ['Sara', 'Noura'] | اسما صاحبتَي الدرجات الأعلى من 80 |
| df["city"].value_counts()["Riyadh"] | 2 | عدد مرات ظهور Riyadh |

## Task 3
تنظيف السجل وتسجيل القرارات

```python
# نسخة مستقلة للتنظيف
clean_df = df.copy()

print("القيم الناقصة في كل عمود:")
print(clean_df.isna().sum())

print("\nعدد الصفوف المكررة:")
print(clean_df.duplicated().sum())

print("\nالطلب المكرر:")
display(clean_df[clean_df.duplicated()])

# حذف التكرار مع الاحتفاظ بأول نسخة
clean_df = clean_df.drop_duplicates().copy()

print("\nعدد الصفوف بعد حذف التكرار:", len(clean_df))
```

```python
# إزالة المسافات وتوحيد كتابة المدن
clean_df["city"] = clean_df["city"].str.strip().str.title()

print("المدن بعد التوحيد:")
print(clean_df["city"].value_counts(dropna=False))

# تصحيح السعر النصي ثم تحويل العمود إلى أرقام
clean_df["price"] = (
    clean_df["price"]
    .replace("eighteen", "18")
    .astype(float)
)

print("\nالأسعار بعد التحويل:")
display(clean_df[["customer", "price"]])

print("نوع عمود السعر:", clean_df["price"].dtype)
```

```python
# عرض التقييمات غير المقبولة
invalid_rating = (
    clean_df["rating"].notna()
    & ~clean_df["rating"].between(1, 5)
)

print("الطلبات ذات التقييم غير المقبول:")
display(clean_df.loc[invalid_rating, ["customer", "rating"]])

# تحويل التقييم غير المقبول إلى قيمة ناقصة
clean_df.loc[invalid_rating, "rating"] = float("nan")

# حساب الوسيط قبل تعويض الكمية
quantity_median = clean_df["quantity"].median()
print("وسيط الكمية:", quantity_median)

print("\nالقيم الناقصة قبل التعبئة:")
print(clean_df.isna().sum())
```

```python
# حساب قيم التعويض وعرضها
quantity_mode = clean_df["quantity"].mode().iloc[0]
rating_median = clean_df["rating"].median()

print("منوال الكمية:", quantity_mode)
print("وسيط التقييم:", rating_median)

# تعبئة القيم الناقصة
clean_df["quantity"] = clean_df["quantity"].fillna(quantity_mode)
clean_df["rating"] = clean_df["rating"].fillna(rating_median)
clean_df["city"] = clean_df["city"].fillna("Unknown")
clean_df["drink"] = clean_df["drink"].fillna("Unknown")

print("\nالقيم الناقصة بعد التعبئة:")
print(clean_df.isna().sum())

print("\nالجدول النظيف:")
display(clean_df)
```

### Decision Log — سجل قرارات التنظيف

| القرار | السبب | عدد الصفوف أو الخلايا المتأثرة |
|---|---|---|
| حذف النسخة المكررة من الطلب 105 | منع احتساب الطلب مرتين | صف واحد |
| إزالة المسافات وتوحيد كتابة المدن | جمع صيغ الرياض تحت اسم واحد | خليتان |
| تصحيح eighteen إلى 18 | السعر مكتوب بالكلمات | خلية واحدة |
| تحويل price إلى float | إتاحة الحسابات العددية | 11 خلية |
| تحويل التقييم 30 إلى قيمة ناقصة | خارج نطاق التقييم من 1 إلى 5 | خلية واحدة |
| تعويض الكمية بالمنوال 1 | وسيط الكمية 1.5، واخترت المنوال لأنه عدد أكواب كامل | خلية واحدة |
| تعويض التقييمات الناقصة بالوسيط 4 | استخدام قيمة وسطية من التقييمات المقبولة | خليتان |
| تعويض المدينة والمشروب بـ Unknown | الحفاظ على الطلب دون تخمين المعلومة | خلية في كل عمود |

القيم المعوّضة تقديرية، وقد تؤثر في الإيرادات ومتوسطات التقييم عند التحليل.

## Task 4
تقرير لصاحب الكافيه

```python
# إيراد كل طلب = السعر × الكمية
clean_df["revenue"] = clean_df["price"] * clean_df["quantity"]

print("إجمالي الإيرادات:", clean_df["revenue"].sum())

print("\nإيرادات المدن من الأعلى إلى الأقل:")
display(
    clean_df.groupby("city")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("\nالكمية المباعة ومتوسط التقييم لكل مشروب:")
display(
    clean_df.groupby("drink").agg(
        total_quantity=("quantity", "sum"),
        average_rating=("rating", "mean")
    ).sort_values("total_quantity", ascending=False)
)

print("\nنسبة الطلبات لكل مدينة (%):")
print(clean_df["city"].value_counts(normalize=True) * 100)

print("\nأعلى طلب من حيث الإيراد:")
display(
    clean_df.loc[
        [clean_df["revenue"].idxmax()],
        ["order_id", "customer", "drink", "revenue"]
    ]
)
```

### توصياتي لصاحب الكافيه
1. أنصح بتوفير مخزون كافٍ لمشروب Latte لأنه الأكثر مبيعًا بكمية 9 أكواب.
2. أنصح بتجربة عرض تسويقي في الرياض لأنها تمثل 45.45% من الطلبات، وتحقق أعلى إيراد بقيمة 138.
3. أنصح بمراجعة جودة Espresso وجمع تقييمات إضافية، لأن متوسط تقييمه 3.5 من 5، وهو الأقل بين المشروبات المعروفة.

## Task 5
وكالة السفر عبر الزمن

```python
trips = {
    "traveler": ["Zara", "Omar", "Lina", "Fahad", "Zara",
                 "Nora", "Yousef", "Salma", "Khalid", "Maya"],
    "departure_year": [2026] * 10,
    "destination_year": ["1450", "3026", "-300", "1999", "1450",
                         "2500", "1000000", "1917", "1450", None],
    "age": [29, 35, 41, 27, 29, 33, 38, -5, 45, 31],
    "price": [12000, 30000, 45000, 8000, 12000,
              22000, 999999, 15000, 12000, 9000],
    "status": ["done", "done", "cancelled", "done", "done",
               "DONE", "done", "done", "Done", "done"]
}

trips_df = pd.DataFrame(trips)

print("سجل الرحلات:")
display(trips_df)

print("\nالقيم الناقصة:")
print(trips_df.isna().sum())

print("\nعدد الصفوف المكررة:")
print(trips_df.duplicated().sum())

print("\nعدد القيم المختلفة في كل عمود:")
print(trips_df.nunique())

print("\nعدد الرحلات لكل حالة:")
print(trips_df["status"].value_counts(dropna=False))
```

```python
# حذف الصف المكرر
trips_clean = trips_df.drop_duplicates().copy()

# توحيد كتابة الحالة
trips_clean["status"] = (
    trips_clean["status"].str.strip().str.lower()
)

# تحويل سنة الوجهة إلى رقم
trips_clean["destination_year"] = pd.to_numeric(
    trips_clean["destination_year"], errors="coerce"
)

print("الرحلات التي بلا وجهة، قبل استبعادها:")
display(
    trips_clean[trips_clean["destination_year"].isna()]
)

trips_clean = trips_clean.dropna(
    subset=["destination_year"]
).copy()

# العمر السالب غير صالح
trips_clean.loc[
    trips_clean["age"] < 0, "age"
] = float("nan")

print("\nالحالات بعد التوحيد:")
print(trips_clean["status"].value_counts())

print("\nالسجل بعد التنظيف الأولي:")
display(trips_clean)
```

```python
# حساب المسافة الزمنية قبل حذف سنة الانطلاق
trips_clean["years_traveled"] = (
    trips_clean["destination_year"]
    - trips_clean["departure_year"]
).abs()

# تحديد اتجاه الرحلة
trips_clean["direction"] = (
    trips_clean["destination_year"]
    > trips_clean["departure_year"]
).map({True: "future", False: "past"})

# حذف سنة الانطلاق لأنها ثابتة في جميع الصفوف
trips_clean = trips_clean.drop(columns=["departure_year"])

print("السجل مع المسافة الزمنية واتجاه الرحلة:")
display(trips_clean)

print("\nأطول رحلة:")
display(
    trips_clean.loc[
        [trips_clean["years_traveled"].idxmax()],
        ["traveler", "destination_year", "years_traveled"]
    ]
)
```

```python
print("أكثر الوجهات طلبًا:")
print(trips_clean["destination_year"].value_counts())

print("\nمتوسط السعر لكل اتجاه:")
print(trips_clean.groupby("direction")["price"].mean())

# الرحلات المكتملة فقط
completed = trips_clean[trips_clean["status"] == "done"]

print("\nعدد الرحلات المكتملة:", len(completed))
print("إيرادات الرحلات المكتملة:", completed["price"].sum())
```

### قرارات تنظيف سجل الرحلات
- حذفت صفًا مكررًا للمسافرة Zara لمنع احتساب الرحلة مرتين.
- وحّدت كتابة الحالة إلى أحرف صغيرة.
- حوّلت سنة الوجهة إلى رقم، واستبعدت رحلة Maya لأن وجهتها ناقصة ولا يمكن حساب اتجاهها أو مسافتها الزمنية.
- حوّلت عمر Salma السالب إلى قيمة ناقصة دون تخمين عمرها، واحتفظت برحلتها لأنها صالحة لتحليل الوجهات والإيرادات.
- احتفظت بالسنة -300 لأنها تمثل وجهة في الماضي ضمن السيناريو.
- احتفظت برحلة Yousef إلى سنة 1000000 بسعر 999999 باعتبارها ممكنة في السيناريو الخيالي، مع اعتبارها قيمة متطرفة تحتاج مراجعة.
