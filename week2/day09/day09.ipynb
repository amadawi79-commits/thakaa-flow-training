## Task 1

أشوف الإعلان مضلّل؛ لأن راتب المدير رفع المتوسط، بينما أغلب الموظفين رواتبهم بين ٦٠٠٠ و٨٠٠٠ ريال.

```python
import numpy as np

salaries = np.array([6000, 6500, 7000, 7200, 7500, 8000, 95000])

print("Mean:", np.mean(salaries))
print("Median:", np.median(salaries))

q1, q3 = np.percentile(salaries, [25, 75])
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = salaries[(salaries < lower) | (salaries > upper)]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower limit:", lower)
print("Upper limit:", upper)
print("Outliers:", outliers)
```

```python
without_manager = salaries[salaries != 95000]

print("Mean without manager:", np.mean(without_manager))
print("Median without manager:", np.median(without_manager))
```

بعد استبعاد راتب المدير، انخفض المتوسط من ١٩٬٦٠٠ إلى حوالي ٧٬٠٣٣ ريال، بينما تغيّر الوسيط قليلًا من ٧٬٢٠٠ إلى ٧٬١٠٠ ريال؛ وهذا يوضح أن المتوسط أكثر تأثرًا بالقيم الشاذة

راتب المدير المرتفع مقارنة بباقي الموظفين يعتبر قيمة شاذة حقيقية، وما نحذفه لمجرد أنه مختلف.

وتكلفة العلاج المرتفعة لحالة معقدة تعتبر قيمة شاذة حقيقية ومهمة للتحليل.

أما تسجيل عمر شخص بـ −٥ سنوات فهو خطأ واضح يحتاج تصحيحًا.

القيمة الشاذة مو دائمًا خطأ؛ لازم نتحقق من سببها قبل حذفها

الـ z-score مثل مقارنة طول شخص بأطوال مجموعته؛ يوضح بُعده عن المتوسط بوحدة الانحراف المعياري.
إذا كان صفرًا فهو عند المتوسط، وإذا كان +٢ فهو أعلى منه بانحرافين معياريين، و−٢ أقل منه بانحرافين.

## Task 2

```python
import numpy as np

np.set_printoptions(suppress=True)

data = np.genfromtxt(
    "insurance.csv",
    delimiter=",",
    skip_header=1,
    usecols=(0, 2, 3, 6)
)

print("Shape:", data.shape)
print("First 3 rows:")
print(data[:3])
```

```python
age = data[:, 0]
bmi = data[:, 1]
children = data[:, 2]
charges = data[:, 3]

print("Columns separated successfully")
```

```python
print(f"{'Column':<12} {'Mean':>12} {'Median':>12} {'Std':>12} {'Min':>12} {'Max':>12}")

for name, values in [
    ("age", age),
    ("bmi", bmi),
    ("children", children),
    ("charges", charges)
]:
    print(
        f"{name:<12}"
        f"{np.mean(values):>12.2f}"
        f"{np.median(values):>12.2f}"
        f"{np.std(values):>12.2f}"
        f"{np.min(values):>12.2f}"
        f"{np.max(values):>12.2f}"
    )
```

```python
print("Missing values per column:", np.isnan(data).sum(axis=0))
print("Total missing values:", np.isnan(data).sum())
```

```python
no_children_count = np.sum(children == 0)
no_children_percent = np.mean(children == 0) * 100

print("Customers without children:", no_children_count)
print(f"Percentage: {no_children_percent:.2f}%")
```

أكبر فرق بين المتوسط والوسيط كان في عمود التكاليف؛ المتوسط حوالي ١٣٬٢٧٠ والوسيط حوالي ٩٬٣٨٢. وهذا يشير إلى أن بعض التكاليف المرتفعة رفعت المتوسط، وأن التوزيع يميل لليمين.

الانحراف المعياري للتكاليف حوالي ١٢٬١٠٥، وهو قريب من المتوسط؛ وهذا يدل على تشتت كبير واختلاف واضح في تكاليف العملاء.

## Task 3

```python
for name, values in [("charges", charges), ("bmi", bmi)]:
    q1, q3 = np.percentile(values, [25, 75])
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    mask = (values < lower) | (values > upper)

    print("\nColumn:", name)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Outlier count:", np.sum(mask))
    print(f"Outlier percentage: {np.mean(mask) * 100:.2f}%")
```

```python
print("Top 5 charges:", np.sort(charges)[-5:][::-1])
print("Top 5 BMI:", np.sort(bmi)[-5:][::-1])
```

```python
for name, values in [("charges", charges), ("bmi", bmi)]:
    z = (values - np.mean(values)) / np.std(values)
    count = np.sum(np.abs(z) > 3)

    print(name, "- Outliers with |z| > 3:", count)
```

اختلف العدد لأن IQR يعتمد على الربيعات، بينما z-score يعتمد على المتوسط والانحراف المعياري. التكاليف المرتفعة رفعت المتوسط والانحراف المعياري، فأصبح حد z-score أوسع وحدّد قيمًا أقل.

```python
q1, q3 = np.percentile(charges, [25, 75])
iqr = q3 - q1

charges_mask = (
    (charges < q1 - 1.5 * iqr) |
    (charges > q3 + 1.5 * iqr)
)

print("Mean with outliers:", np.mean(charges))
print("Mean without outliers:", np.mean(charges[~charges_mask]))
```

```python
smoker = np.genfromtxt(
    "insurance.csv",
    delimiter=",",
    skip_header=1,
    usecols=4,
    dtype=str,
    encoding="utf-8"
)

print("Smokers among outliers:",
      f"{np.mean(smoker[charges_mask] == 'yes') * 100:.2f}%")

print("Mean charges - smokers:",
      np.mean(charges[smoker == "yes"]))

print("Mean charges - non-smokers:",
      np.mean(charges[smoker == "no"]))
```

قراري أحتفظ بالقيم الشاذة وأتحقق منها قبل حذفها؛ لأن اختلافها لا يعني أنها أخطاء. وجدت أن 97.84% من أصحاب التكاليف الشاذة مدخنون، ومتوسط تكاليف المدخنين حوالي 32,050 مقابل 8,434 لغير المدخنين، وهذا يشير إلى ارتباط التدخين بارتفاع التكاليف في هذه البيانات.

حذف هذه القيم قد يفقدنا حالات حقيقية ومهمة لتدريب النموذج. كذلك انخفض متوسط التكاليف من 13,270 إلى 9,928 عند استبعادها، مما يوضح تأثيرها الكبير على المتوسط.

## Task 5

```python
rng = np.random.default_rng(7)
brightness = 1000 + rng.normal(0, 1, 240)

for start in [30, 102, 174]:
    brightness[start:start + 5] -= 20

print("Mean brightness:", np.mean(brightness))
print("Median brightness:", np.median(brightness))
```

```python
hours = np.arange(240)

z = (brightness - np.mean(brightness)) / np.std(brightness)
transit_mask = z < -3

print("Number of low readings:", np.sum(transit_mask))
print("Hours of low brightness:", hours[transit_mask])
```

```python
low_hours = hours[transit_mask]

transit_starts = low_hours[
    np.r_[True, np.diff(low_hours) > 1]
]

period_hours = np.diff(transit_starts)
period_days = period_hours / 24

print("Transit start hours:", transit_starts)
print("Period in hours:", period_hours)
print("Period in days:", period_days)
```

```python
normal_brightness = np.mean(brightness[~transit_mask])
transit_brightness = np.mean(brightness[transit_mask])

drop = normal_brightness - transit_brightness
drop_percent = (drop / normal_brightness) * 100

print("Brightness drop:", drop)
print(f"Drop percentage: {drop_percent:.2f}%")
```
