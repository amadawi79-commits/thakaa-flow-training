### 1. لماذا نرسم البيانات؟

قد تتشابه مجموعات البيانات في المتوسط والانحراف والارتباط، لكن عند رسمها تظهر اختلافات مثل القيم الشاذة أو وجود علاقة منحنية. لذلك لا يكفي الاعتماد على الأرقام الملخصة وحدها.

مثال: متوسط مذاكرتي ساعتان يوميًا قد يعني أنني أذاكر ساعتين كل يوم، أو أربع ساعات في يوم ولا أذاكر في اليوم التالي. المتوسط نفسه، لكن الانتظام مختلف.

### 2. اختيار الرسمة المناسبة

| السؤال | الرسمة | السبب |
|---|---|---|
| كيف تغيرت مبيعات المقهى خلال 12 شهرًا؟ | Line — خطية | توضح تغير المبيعات مع الوقت. |
| أي فرع باع أكثر هذا الشهر من 5 فروع؟ | Bar — أعمدة | تسهّل مقارنة مبيعات الفروع. |
| كيف تتوزع أعمار 500 عميل؟ | Histogram — مدرج تكراري | يوضح عدد العملاء في كل فئة عمرية. |
| هل يزيد الإيجار كلما زادت مساحة الشقة؟ | Scatter — مبعثرة | توضح العلاقة بين المساحة والإيجار. |
| هل في رواتب الموظفين قيم شاذة؟ | Box plot — صندوقية | تساعد على اكتشاف الرواتب البعيدة عن بقية القيم. |

### 3. أجزاء الرسمة

- Figure: الشكل الكامل الذي يحتوي على رسمة واحدة أو أكثر.
- Axes: منطقة الرسم داخل الشكل، وفيها تُعرض البيانات والمحاور.
- العنوان: يوضح موضوع الرسمة.
- عنوانا المحورين: يوضحان ما تمثله القيم ووحدة القياس، مثل الشهر والمبيعات بالريال.
- Legend: يوضح معنى كل لون أو رمز عند وجود أكثر من مجموعة بيانات.

بدون عنوان قد لا يفهم المدير موضوع الرسمة، وبدون أسماء المحاور قد لا يعرف معنى الأرقام أو وحدتها، وبدون Legend قد يخلط بين المجموعات.

### 4. رسمة مضللة

الخدعة هي أن محور النسب يبدأ من قيمة قريبة من 94% بدل الصفر، فيبدو عمود 96% كأنه ضعف عمود 94%، رغم أن الفرق نقطتان مئويتان فقط.

أصلح الرسمة بجعل محور الأعمدة يبدأ من صفر وينتهي عند 100%، وكتابة النسبة فوق كل عمود، حتى يظهر الفرق بحجمه الصحيح.

```python
import matplotlib.pyplot as plt

days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
latte = [42, 38, 35, 40, 44, 58, 61]
espresso = [30, 28, 33, 31, 29, 35, 22]

drinks = ["Latte", "Espresso", "Mocha", "Tea"]

ratings = [
    5, 4, 4, 5, 3, 5, 4, 2, 5, 4,
    4, 5, 3, 4, 5, 5, 1, 4, 4, 5,
    3, 4, 5, 4, 2, 5, 4, 4, 3, 5,
    4, 5, 5, 4, 3, 4, 5, 4, 4, 5
]

temp = [18, 22, 25, 27, 30, 33, 35, 38, 40, 41]
iced = [5, 9, 12, 15, 20, 26, 30, 35, 39, 38]
```

```python
plt.figure(figsize=(9, 5))

plt.plot(days, latte, marker="o", label="Latte")
plt.plot(days, espresso, marker="s", label="Espresso")

plt.title("Daily Latte and Espresso Sales")
plt.xlabel("Day")
plt.ylabel("Cups Sold")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

أعلى مبيعات اللاتيه كانت يوم الجمعة بـ61 كوبًا. أما الإسبريسو فانخفضت مبيعاته من 35 كوبًا يوم الخميس إلى 22 كوبًا يوم الجمعة، وهو أقل عدد له خلال الأسبوع.

```python
totals = [sum(latte), sum(espresso), 145, 96]

plt.figure(figsize=(8, 5))
plt.bar(drinks, totals, color=["tan", "brown", "chocolate", "seagreen"])

for i, value in enumerate(totals):
    plt.text(i, value + 3, str(value), ha="center")

plt.title("Weekly Sales by Drink")
plt.xlabel("Drink")
plt.ylabel("Cups Sold")
plt.ylim(0, max(totals) + 40)
plt.tight_layout()
plt.show()
```

اللاتيه هو الأكثر مبيعًا بإجمالي 318 كوبًا، يليه الإسبريسو بـ208 أكواب، ثم الموكا بـ145 كوبًا، والشاي الأقل بـ96 كوبًا.

أتوقع أن يكون التقييم 4 هو الأكثر تكرارًا.

```python
plt.figure(figsize=(8, 5))

plt.hist(
    ratings,
    bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5],
    color="steelblue",
    edgecolor="black"
)

plt.title("Customer Ratings Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Ratings")
plt.xticks([1, 2, 3, 4, 5])
plt.tight_layout()
plt.show()

for rating in range(1, 6):
    print(f"Rating {rating}: {ratings.count(rating)}")
```

توقعي كان صحيحًا؛ التقييم 4 هو الأكثر تكرارًا بـ17 مرة، يليه التقييم 5 بـ15 مرة. معًا يمثلان 80% من التقييمات، مما يشير إلى رضا أغلب العملاء.

```python
plt.figure(figsize=(8, 5))

plt.scatter(temp, iced, color="teal")

plt.title("Temperature and Iced Coffee Sales")
plt.xlabel("Temperature (C)")
plt.ylabel("Iced Coffee Cups Sold")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

تظهر علاقة طردية واضحة؛ ترتفع مبيعات القهوة المثلجة عمومًا مع ارتفاع الحرارة، مع انخفاض بسيط من 39 إلى 38 كوبًا عند الانتقال من 40 إلى 41 درجة.

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

# Line
axes[0, 0].plot(days, latte, marker="o", label="Latte")
axes[0, 0].plot(days, espresso, marker="s", label="Espresso")
axes[0, 0].set_title("Daily Sales")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Cups Sold")
axes[0, 0].legend()

# Bar
axes[0, 1].bar(drinks, totals, color="tan")
axes[0, 1].set_title("Weekly Sales by Drink")
axes[0, 1].set_xlabel("Drink")
axes[0, 1].set_ylabel("Cups Sold")
axes[0, 1].set_ylim(0, max(totals) + 40)

for i, value in enumerate(totals):
    axes[0, 1].text(i, value + 3, str(value), ha="center")

# Histogram
axes[1, 0].hist(
    ratings,
    bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5],
    color="steelblue",
    edgecolor="black"
)
axes[1, 0].set_title("Customer Ratings")
axes[1, 0].set_xlabel("Rating")
axes[1, 0].set_ylabel("Number of Ratings")
axes[1, 0].set_xticks([1, 2, 3, 4, 5])

# Scatter
axes[1, 1].scatter(temp, iced, color="teal")
axes[1, 1].set_title("Temperature and Iced Coffee Sales")
axes[1, 1].set_xlabel("Temperature (C)")
axes[1, 1].set_ylabel("Cups Sold")

fig.suptitle("Cafe Dashboard", fontsize=18)
fig.tight_layout(rect=[0, 0, 1, 0.95])
fig.savefig("cafe_dashboard.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("insurance.csv")

print("Rows before:", len(df))
print("Duplicates:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Rows after:", len(df))
display(df.head())
```

```python
mean_charge = df["charges"].mean()
median_charge = df["charges"].median()

plt.figure(figsize=(9, 5))

plt.hist(df["charges"], bins=30, color="skyblue", edgecolor="black")

plt.axvline(
    mean_charge, color="red", linestyle="--",
    label=f"Mean: {mean_charge:,.2f}"
)
plt.axvline(
    median_charge, color="green", linestyle="--",
    label=f"Median: {median_charge:,.2f}"
)

plt.title("Medical Insurance Charges Distribution")
plt.xlabel("Charges")
plt.ylabel("Number of People")
plt.legend()
plt.tight_layout()
plt.show()

print(f"Mean: {mean_charge:,.2f}")
print(f"Median: {median_charge:,.2f}")
```

تتركز أغلب التكاليف في القيم المنخفضة، مع ذيل طويل نحو التكاليف المرتفعة. المتوسط أعلى من الوسيط بسبب تأثره بالقيم العالية، لذلك الوسيط أقرب لتمثيل التكلفة المعتادة.

```python
plt.figure(figsize=(10, 6))

for status, color, label in [
    ("no", "steelblue", "Non-smoker"),
    ("yes", "tomato", "Smoker")
]:
    group = df[df["smoker"] == status]

    plt.scatter(
        group["age"],
        group["charges"],
        color=color,
        label=label,
        alpha=0.6,
        s=20
    )

plt.title("Age and Medical Charges by Smoking Status")
plt.xlabel("Age (years)")
plt.ylabel("Charges")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

تظهر ثلاثة شرائط رئيسية تقريبًا، مع بعض النقاط المتفرقة. الشريط الأقل تكلفة يغلب عليه غير المدخنين، بينما يغلب المدخنون على الشريط الأعلى. وتميل التكاليف إلى الارتفاع مع العمر داخل الشرائط.

```python
smoking_mean = df.groupby("smoker")["charges"].mean()
region_mean = df.groupby("region")["charges"].mean()

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

smoking_mean.plot.bar(
    ax=axes[0], color=["steelblue", "tomato"], rot=0
)
axes[0].set_title("Average Charges by Smoking Status")
axes[0].set_xlabel("Smoker")
axes[0].set_ylabel("Average Charges")

region_mean.plot.bar(
    ax=axes[1], color="teal", rot=15
)
axes[1].set_title("Average Charges by Region")
axes[1].set_xlabel("Region")
axes[1].set_ylabel("Average Charges")

fig.tight_layout()
plt.show()

print("Average charges by smoking status:")
print(smoking_mean.round(2))

print("\nAverage charges by region:")
print(region_mean.round(2))
```

متوسط تكاليف المدخنين 32,050.23، مقابل 8,440.66 لغير المدخنين، أي نحو 3.8 أضعاف. وأعلى متوسط بين المناطق في southeast بقيمة 14,735.41، وأقل متوسط في southwest بقيمة 12,346.94.

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax in axes:
    ax.bar(region_mean.index, region_mean.values, color="teal")
    ax.set_xlabel("Region")
    ax.set_ylabel("Average Charges")
    ax.tick_params(axis="x", rotation=15)

axes[0].set_title("Truncated Axis: Starts at 12,000")
axes[0].set_ylim(12000, region_mean.max() + 1000)

axes[1].set_title("Full Axis: Starts at Zero")
axes[1].set_ylim(0, region_mean.max() + 1000)

fig.tight_layout()
plt.show()

highest = region_mean.idxmax()
lowest = region_mean.idxmin()
ratio = region_mean.max() / region_mean.min()

print(region_mean.round(2))
print(f"\nHighest region: {highest}")
print(f"Lowest region: {lowest}")
print(f"Highest / lowest: {ratio:.2f} times")
print(f"Percentage higher: {(ratio - 1) * 100:.2f}%")
```

```python
import numpy as np

rng = np.random.default_rng(2026)
sol = np.arange(1, 61)

temp = np.round(
    -60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60),
    1
)

storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1

temp[storm == 1] -= 12

wind = np.round(
    20 + storm * 35 + rng.normal(0, 5, 60),
    1
)
```

```python
print(f"Mean: {mean_charge:,.2f}")
print(f"Median: {median_charge:,.2f}")
```

```python
coldest = np.argmin(temp)

plt.figure(figsize=(12, 5))

plt.plot(sol, temp, label="Temperature", color="steelblue")

plt.scatter(
    sol[storm == 1],
    temp[storm == 1],
    color="red",
    label="Storm",
    zorder=3
)

plt.axhline(
    temp.mean(),
    color="green",
    linestyle="--",
    label=f"Mean: {temp.mean():.1f} C"
)

plt.annotate(
    f"Coldest: Sol {sol[coldest]}\n{temp[coldest]:.1f} C",
    xy=(sol[coldest], temp[coldest]),
    xytext=(15, 35),
    textcoords="offset points",
    arrowprops={"arrowstyle": "->"}
)

plt.title("Mars Temperature over 60 Sols")
plt.xlabel("Sol (Martian Day)")
plt.ylabel("Temperature (C)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Coldest Sol: {sol[coldest]}")
print(f"Lowest temperature: {temp[coldest]:.1f} C")
print(f"Mean temperature: {temp.mean():.1f} C")
```

تتغير الحرارة خلال الأيام الستين، وكان أبرد يوم هو اليوم 48 بدرجة −87.5، وهو من أيام العواصف. متوسط الحرارة خلال الفترة نحو −60.7 درجة.

```python
fig, axes = plt.subplots(
    2, 1, figsize=(12, 8), sharex=True
)

axes[0].plot(sol, temp, color="steelblue")
axes[0].scatter(
    sol[storm == 1], temp[storm == 1],
    color="red", label="Storm", zorder=3
)
axes[0].set_title("Mars Temperature")
axes[0].set_ylabel("Temperature (C)")
axes[0].legend()
axes[0].grid(alpha=0.3)

colors = np.where(storm == 1, "red", "gray")
axes[1].bar(sol, wind, color=colors)
axes[1].set_title("Mars Wind — Red Bars Indicate Storms")
axes[1].set_xlabel("Sol (Martian Day)")
axes[1].set_ylabel("Wind Speed")
axes[1].grid(axis="y", alpha=0.3)

fig.tight_layout()
plt.show()
```

تظهر أيام العواصف بالأحمر في الرسمين، وتتزامن مع ارتفاع سرعة الرياح وانخفاض الحرارة مقارنة بالأيام القريبة منها. مشاركة محور الأيام تسهّل مقارنة التغيرين.

أتوقع أن تكون أفضل نافذة في الفترة الدافئة قرب بداية البيانات، بشرط أن تكون الأيام الخمسة بلا عواصف. سأتحقق بمقارنة متوسط حرارة جميع النوافذ المناسبة.

```python
best_start = None
best_mean = -np.inf

for start in range(len(sol) - 4):
    window_storm = storm[start:start + 5]
    window_temp = temp[start:start + 5]

    if np.all(window_storm == 0):
        average = window_temp.mean()

        if average > best_mean:
            best_mean = average
            best_start = start

best_days = sol[best_start:best_start + 5]

print(f"Best window: Sol {best_days[0]} to {best_days[-1]}")
print(f"Mean temperature: {best_mean:.2f} C")
print("Temperatures:", temp[best_start:best_start + 5])
print("Storm flags:", storm[best_start:best_start + 5])
```
